```python
import torch
import os
import matplotlib.pyplot as plt
from PIL import Image

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[+] Execution Device: {device}")
print(f"[+] Local Repository State:")
print(f"    - challenge_data/ folder present: {os.path.exists('./challenge_data')}")
print(f"    - resnet50_phaseA.pt weights found: {os.path.exists('resnet50_phaseA.pt')}")
print(f"    - data/synthetic_target/ populated: {os.path.exists('./data/synthetic_target/') and len(os.listdir('./data/synthetic_target/')) > 0}")
```

## Phase A: Baseline Evaluation
Here we calculate the baseline metrics and compute the initial Domain Shift Penalty ($\Delta shift$).

```python
from data_loader import get_dataloaders
from classifier import build_resnet50
from domain_adaptation import evaluate_domain_shift

dataloaders, class_names = get_dataloaders(batch_size=32)
model = build_resnet50(num_classes=6).to(device)
model.load_state_dict(torch.load('resnet50_phaseA.pt', map_location=device))

print("=== RUNNING BASELINE METRIC EXTRACTION ===")
acc_src, acc_tgt, delta = evaluate_domain_shift(model, dataloaders, device)
print(f"\n[RESULTS] Source (Real) Acc: {acc_src:.4f} | Target (Sketch) Acc: {acc_tgt:.4f}")
print(f"[RESULTS] Baseline Domain Shift Penalty (Delta): {delta:.4f}")
```

## Phase B: Neural Style Transfer Showcase
Visual verification of the style-transfer generation pipeline. Let's render a sample triple (Content, Style, Generated Synthesized Output).

```python
synth_dir = './data/synthetic_target/'
if os.path.exists(synth_dir) and len(os.listdir(synth_dir)) > 0:
    sample_cls = os.listdir(synth_dir)[0]
    sample_img_path = os.path.join(synth_dir, sample_cls, os.listdir(os.path.join(synth_dir, sample_cls))[0])

    img = Image.open(sample_img_path)
    plt.figure(figsize=(5,5))
    plt.imshow(img)
    plt.title(f"Synthetic Target Sample (Class: {sample_cls})")
    plt.axis('off')
    plt.show()
else:
    print("[-] Directory data/synthetic_target/ empty. Run your generate_synthetic.py script first.")
```
